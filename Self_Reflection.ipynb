```python
##What will be summarized

source_text = """ A retailer introduced a customer support chatbot. During a three-month pilot, average response time decreased from 10 minutes to 2 minutes. Customer satisfaction increased from 78% to 86%. Complex billing cases still required human agents, and long-term customer retention was not measured. """

print(source_text)
```

Summary Prompt

Summarize the passage in only two or three sentences.

Requirements:
- Use a professional tone
- Write for a business audience
- Include the main action
- Include measurable results
- Include one limitation
- Do not add unsupported information

```python
initial_summary = """
The retailer’s chatbot improved customer service by making
responses faster and increasing customer satisfaction. Some
complicated cases still required human support.
"""

print(initial_summary)
```

Self-Reflection Prompt:

Review the summary using these requirements:

1. Accuracy: Are all claims supported by the source?
2. Completeness: Are the main results and limitation included?
3. Clarity: Is it easy for a business audience to understand?
4. Length: Is it two or three sentences?
5. Format: Is it one professional paragraph?
6. Focus: Does it avoid unnecessary information?

Identify what is missing or unclear.
Then provide specific revision instructions.
Do not rewrite the summary yet.

This prompt allows for the summary to be double checked for the best output.

```python
critique = """
The summary is generally accurate, but it does not include the
specific numerical results. It should state that response time
decreased from 10 minutes to 2 minutes and satisfaction increased
from 78% to 86%. It also needs to mention that long-term retention
was not measured. The revised version should remain two or three
sentences and use a professional business tone.
"""

print(critique)
```

Improvement Prompt:

Rewrite the summary using the critique.

Requirements:
- Use two or three sentences
- Use a professional business tone
- Include all important numbers
- Include the limitation
- Preserve the original meaning
- Do not add unsupported information
- Use one cohesive paragraph

This reiterates some of the intial prompt while ensuring a better output.

```python
required_details = [
    "10 minutes",
    "2 minutes",
    "78%",
    "86%",
    "long-term customer retention"
]

for detail in required_details:
    assert detail in improved_summary

print("QUALITY CHECK: PASS")
```

```python
improved_summary = """
During a three-month pilot, the retailer’s customer-support
chatbot reduced average response time from 10 minutes to 2 minutes
and increased customer satisfaction from 78% to 86%. Complex billing
cases were escalated to human agents, and long-term customer retention was not measured.
"""

print(improved_summary)
```
