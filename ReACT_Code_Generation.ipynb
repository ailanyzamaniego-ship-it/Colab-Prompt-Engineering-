```python
sales_data = [
    {"product": "Notebook", "units": 4, "price": 5.00},
    {"product": "Pen", "units": 10, "price": 1.50},
    {"product": "Notebook", "units": 2, "price": 5.00},
    {"product": "Backpack", "units": 1, "price": 30.00},
]

sales_data
```

Plan:
1. Multiply units by price for each sale.
2. Group repeated products.
3. Add the revenue for each product.
4. Check for empty data and invalid negative values.

```python
from collections import defaultdict

def calculate_revenue(sales):
    if not sales:
        return "No sales records available."

    revenue_by_product = defaultdict(float)

    for row in sales:
        if row["units"] < 0 or row["price"] < 0:
            return "Error: Units and prices cannot be negative."

        revenue_by_product[row["product"]] += (
            row["units"] * row["price"]
        )

    total_revenue = sum(revenue_by_product.values())

    print(f"Total revenue: ${total_revenue:.2f}")
    print("\nRevenue by product:")

    for product, revenue in sorted(revenue_by_product.items()):
        print(f"{product}: ${revenue:.2f}")

    return revenue_by_product

calculate_revenue(sales_data)
```

Observation:
The code correctly calculates total revenue and combines the two Notebook records. The total is $75.00.

```python
## Test edge case
empty_sales_data = []

calculate_revenue(empty_sales_data)
```
