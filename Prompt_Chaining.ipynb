```python
## Sample issues
customer_message = """
I was charged three times for my monthly subscription.
Please refund the duplicate charges.
"""

print(customer_message)
```

Prompt 1: You are a customer-support intent classifier.

Classify the customer message as:
- billing
- technical support
- account access
- general question

Return:
1. The category
2. A short explanation
3. The urgency level

Customer message:
{customer_message}

Explanation: This prompt will help determine what exactly the customer needs assistance on and will categorize it by urgency. This will help determine how soon this issue should be resolved and provide a summary.

```python
classification = {
    "category": "billing",
    "explanation": "The customer reports a duplicate subscription charges.",
    "urgency": "medium"
}

print(classification)
```

Prompt 2: Identify Missing Information

You are a customer-support information-gathering assistant.

Using the customer message and classification, identify the
information needed to investigate the problem.

Ask for no more than two pieces of information.
Do not request passwords or full payment-card numbers.

Customer message:
{customer_message}

Classification:
{classification}
Explanation: This prompt will help determine what additional information is needed to resolve the issue. A limit is put on the amount of information that can be needed. It also makes sure that extremly personal data is not asked for.

```python
missing_information = [
    "The date of the duplicate charge",
    "The last four digits of the payment method"
]

print("Information needed:")
for item in missing_information:
    print("-", item)
```

Prompt 3: Write the Customer Response

You are a helpful customer-support representative.

Write a concise, empathetic response that:
- Acknowledges the customer’s concern
- Explains the next step
- Does not promise a refund before verification
- Does not request sensitive information
- Uses a professional tone

Use the customer message, classification, and missing information.

Explanation: This prompt ensures that the response is professional and updates the customer on what will happen next. It also ensures that no promises are made regarding the resoultion and does not ask any questions that can potentially be a privacy risk.

```python
response = """
I’m sorry about the duplicate charge. We can review this for you.
Please provide the transaction date and the last four digits of the
payment method through the secure support form. Our billing team
will verify the charge and determine the appropriate next step.
"""

print(response)
```

Prompt 4: Decide Whether to Escalate

You are a support escalation assistant.

Escalate the issue if:
- The customer requests a refund
- Multiple charges are reported
- The transaction cannot be verified

Return:
1. AUTOMATE or ESCALATE
2. One-sentence reason

Explanation: This prompt ensures that the issues get escalated under certain circumstances.

```python
escalation_decision = {
    "decision": "ESCALATE",
    "reason": "The customer reports a duplicate charges and requests a refund."
}

print(escalation_decision)
```
